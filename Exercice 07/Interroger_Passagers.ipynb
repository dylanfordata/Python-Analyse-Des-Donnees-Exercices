```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
```

```python
nb_moins_18 = titanic["age"].lt(18).sum()
print("Moins de 18 ans :", nb_moins_18)
```

```python
print(
    titanic[["fare", "class", "embark_town"]]
    .sort_values("fare", ascending=False)
    .head(5)
)
```

```python
premiere = titanic[titanic["class"] == "First"]
taux_femmes = (premiere["sex"] == "female").mean() * 100
print(f"Femmes en 1re classe : {taux_femmes:.1f} %")
```

```python
femmes_1 = titanic[
    (titanic["class"] == "First") &
    (titanic["sex"] == "female")
]
hommes_3 = titanic[
    (titanic["class"] == "Third") &
    (titanic["sex"] == "male")
]

print(f"Survie femmes 1re : {femmes_1['survived'].mean()*100:.1f} %")
print(f"Survie hommes 3e : {hommes_3['survived'].mean()*100:.1f} %")
```

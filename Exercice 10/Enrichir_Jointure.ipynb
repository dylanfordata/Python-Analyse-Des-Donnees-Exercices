```python
import pandas as pd
import seaborn as sns
titanic = sns.load_dataset("titanic")

print(titanic.head())

ports = pd.DataFrame({
    "embark_town": [
        "Southampton", "Cherbourg", "Queenstown"
    ],
    "pays": [
        "Angleterre", "France", "Irlande"
    ]
})

titanic = titanic.merge(
    ports,
    on="embark_town",
    how="left"
)

print(titanic[["embark_town", "pays"]].head())

print("\nNombre de passagers par pays :")
print(titanic["pays"].value_counts())

print("\nTaux de survie par pays :")
print(titanic.groupby("pays")["survived"].mean() * 100)
```
